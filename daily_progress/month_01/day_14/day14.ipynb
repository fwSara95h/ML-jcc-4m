### MC14.1 : Two Sum (O(N))


> Goal : Find two numbers in a list that add up to Target.


````python
def two_sum(nums: list[int], target: int) -> tuple[int, int] | None:
    seen: dict[int, int] = {}

    for i, value in enumerate(nums):
        needed = target - value
        if needed in seen:
            return seen[needed], i
        seen[value] = i

    return None
````

````python
nums = [2, 7, 11, 15]
target = 9

result = two_sum(nums, target)
print(result)
````

> **Deep Dive:** Naive: Nested Loops *(O(N²))*. Pro: Use a Dictionary (Seen Map). Calculate `needed = target - current`. If `needed` is in `map`, you win.

### MC14.2 : The Palindrome (Slicing)


> Goal : Check if a string is a palindrome (ignoring space/case).


````python
def is_palindrome(text: str) -> bool:
    cleaned = "".join(ch.lower() for ch in text if ch.isalnum())
    return cleaned == cleaned[::-1]
````

````python
print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("hello"))
````

> **Deep Dive:** Clean string → `s == s[::-1]`. Slicing creates a reversed copy in memory.


````python
## memory behavior check

s = "abcd"
rev = s[::-1]

print(id(s) == id(rev))
````

### MC14.3 : Anagrams (Frequency)


> Goal : Are "silent" and "listen" anagrams?


````python
from collections import Counter

def are_anagrams(a: str, b: str) -> bool:
    return Counter(a) == Counter(b)
````

````python
print(are_anagrams("silent", "listen"))
print(are_anagrams("apple", "papel"))
print(are_anagrams("apple", "appeal"))
````

> **Deep Dive:** 
> 1. Sort both (O(N log N)).
> 2. Count frequencies: `{ 's': 1, 'i': 1... }` (O(N)).
> Use `collections.Counter`.

### MC14.4 : Flattening (Recursion)


> Goal : Flatten `[1, [2, [3, 4]]]` into `[1, 2, 3, 4]`.


````python
def flatten(items):
    for item in items:
        if isinstance(item, list):
            yield from flatten(item)
        else:
            yield item
````

````python
nested = [1, [2, [3, 4]], 5]
flat = list(flatten(nested))

print(flat)
````

> **Deep Dive:** Write a recursive function. If item is list → recurse. Else → yield.

### MC14.5 : FizzBuzz (The Logic Gate)


> Goal : Print 1-100. Div by 3 "Fizz", by 5 "Buzz", both "FizzBuzz".


````python
def fizzbuzz(starter=1, ender=100):
    for i in range(starter, ender+1):
        if i % 15 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)
````

````python
fizzbuzz(1, 30)
````

> **Deep Dive:** Order matters! Check `% 15` (both) FIRST. If you check 3 first, "FizzBuzz" numbers will just print "Fizz".


````python
## One liner fizzbuzz
fz, bz = "Fizz", "Buzz"
fzbz = lambda n: fz+bz if n % 15 == 0 else fz if n % 3 == 0 else bz if n % 5 == 0 else str(n)
print('\n'.join([fzbz(i) for i in range(37, 46)]))
````

````python
## Another alternative FizzBuzz implementation
def fizzbuzz_alt(starter=1, ender=100):
    fizzbuzzdict = {3: "Fizz", 5: "Buzz"}
    for i in range(starter, ender+1):
        output = "".join(v for k, v in fizzbuzzdict.items() if i % k == 0)  
        print(output or i)

fizzbuzz_alt(83, 92)
````

### MC14.6 : Deduplication (O(N))


> Goal : Remove duplicates from a list while keeping order.


````python
def dedupe_preserve_order(items: list[int]) -> list[int]:
    seen = set()
    result = []

    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)

    return result
````

````python
data = [1, 2, 1, 3, 2, 4]
print(dedupe_preserve_order(data))
````

> **Deep Dive:** `list(set(x))` destroys order. Solution: Iterate, check if item in seen_set, if not → append to result.

### MC14.7 : Binary Search (O(log N))


> Goal : Find index of number in sorted list.


````python
def binary_search(nums: list[int], target: int) -> int | None:
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return None
````

````python
nums = [1, 3, 5, 7, 9, 11]
print(binary_search(nums, 7))
print(binary_search(nums, 2))
````

> **Deep Dive:** Check middle. If too low, eliminate left half. If too high, eliminate right half. Repeat. Much faster than scanning.

### MC14.8 : Missing Number (Math)


> Goal : List 1 to 100 is missing one number. Find it.


````python
def find_missing(nums: list[int], n: int) -> int:
    expected = n * (n + 1) // 2
    actual = sum(nums)
    return expected - actual
````

````python
nums = list(range(1, 101))
nums.remove(73)

print(find_missing(nums, 100))
````

> **Deep Dive:** Don’t scan. Calculate Expected Sum `N(N+1)/2`. Subtract Actual Sum. The difference is the missing number.

### MC14.9 : Grouping (itertools)


> Goal : Group list of dicts by "category".


````python
from itertools import groupby
from operator import itemgetter

def group_by_category(items: list[dict]) -> dict[str, list[dict]]:
    items_sorted = sorted(items, key=itemgetter("category"))
    grouped = {}

    for key, group in groupby(items_sorted, key=itemgetter("category")):
        grouped[key] = list(group)

    return grouped
````

````python
data = [
    {"name": "apple", "category": "fruit"},
    {"name": "carrot", "category": "veg"},
    {"name": "banana", "category": "fruit"},
]

group_by_category(data)
````

> **Deep Dive:** Use `itertools.groupby`.
> **Note:** Data MUST be sorted by the key first!

### MC14.10 : Merge Sorted Lists


> Goal : Combine two sorted lists into one sorted list (O(N)).


````python
def merge_sorted(a: list[int], b: list[int]) -> list[int]:
    i = j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    result.extend(a[i:])
    result.extend(b[j:])
    return result
````

````python
a = [1, 3, 5, 7]
b = [2, 4, 6, 8]

print(merge_sorted(a, b))
````

> **Deep Dive:** Two Pointers. Compare head of A and head of B. Take smaller. Move pointer. Repeat.
